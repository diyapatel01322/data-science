```python
#tuples are using parentheses()
no=(10,20,30,40)
print(no)
```

```python
#tuple with different data type
#without parentheses
no=10,20,30,40
print(no)
```

```python
#tuple indexing
no=(10,20,30,40,50)
print(no[0])
print(no[2])
print(no[-1])
```

```python
#tuple method
#count
no=(10,20,10,30,45,10)
print=(no.count(10))
```

```python
#count
numbers=(10,20,30,10,40)
print(numbers.count(10))
```

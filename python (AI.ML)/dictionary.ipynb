```python
student ={
    "name":"rahul",
    "age":21,
    "course":"python"
}
print(student)
```

```python
#Accese values
print(student["name"])
print(student["course"])
```

```python
#get()
print(student.get("name"))
print(student.get("course"))
```

```python
#adding a new key value
student["course"]="css"
print(student)
```

```python
#updating values
student["age"]=23
print(student)
```

```python
#update()
student={
    "name":"rahul",
    "age":23
}
student.update({
        "age":21,
        "city":"gandhinagar"
    })
print(student)
```

```python
#remove elements
student.pop("age")
print(student)
```

```python
#popitem()
removed=student["city"]
print(student)
```

```python
#clesr
student.clear()
print(student)
```

```python
#dictination method keys()
# runs all key
student={
    "name":"rahul",
    "age":21,
    "course":"python"
}
print(student.keys())
```

```python
#values
# runs all values
print(student.values())
```

```python
#items()
# Run key-values pair
print(student.items())
```

```python
#Loop Through Dictionary
student={
    "name":"rahul",
    "age":21,
    "course":"python"
}
for key in student:
    print(key)
```

```python
for values in student.values():
    print(values)
```

```python
#Key and values
for key, value in student.items():
    print(key,".",value)
```

```python
#dictionary length
student={
    "name":"rahul",
    "age":21,
    "course":"python"
}
print(len(student))
```

```python
#diffrent data type
person={
    "name":"rahul",
    "age":21, 
    "is_student":True,
    "skills":["python","html","css"]
}
print(person)
```

```python
#dictionary with list
student={
    "name":"rahul",
    "marks":[80,85,70,75,90]
}
print(student["marks"])
print(student["marks"][2])
```

```python
#Duplicate key 
student={
    "name":"rahul",
    "age":21,
    "age":22
}
print(student)
```

```python
#dictionary keys
student={
    "name":"rahul",
    101:"python",
    (1,2):"
    "Tuple"
}
print(student)
```

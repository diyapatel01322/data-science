```python
no={10,20,30,40,50}
print(no)
```

```python
#duplicate number
no={10,20,25,20,10,30}
print(no)
```

```python
#Empty set
x=set()
print(x)
```

```python
#Set method
#Add
no={10,20,30,40}
no.add(45)
print(no)
```

```python
# update
no.update([40,50,55])
print(no)
```

```python
#remove()
no.remove(45)
print(no)
```

```python
#discard()
no.discard(40)
print(no)
```

```python
#pop()
remove=no.pop ()
print("remove:",remove)
print(no)
```

```python
#clear()
no.clear()
print(no)
```

```python
#set operations
#union
a={1,2,3}
b={4,5,6}
result=a|b
print(result)
```

```python
#intersection
#return commom elements
a={1,2,3}
b={3,5,6}
result=a&b
print(result)
```

```python
#difference
#return elements present in the first set but not the second
print(a-b)
```

```python
#symmetric difference ^
print(a ^ b)
```

```python
#set membership 
#you can use "in" and " not in ".
no={10,20,30,40}
print(20 in no)
print(50 in no)
```

```python
#loop through set
for no in no:
    print(no)
```

```python
#set comparison
a={1,2}
b={1,2,3,4}
print(a.issubset(b))
```

```python
#isdisjoint()
a={1,2,3}
b={4,5,6}
print(a.isdisjoint(b))
```

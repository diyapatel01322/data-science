```python
#multiple return values
def cal(a,b):
    addition=a+b
    subtraction =a-b
    return addition,subtraction
x,y=cal(20,10)
print("addition:",x)
print("subtraction:",y)
```

```python
#function with list
def find_sum(no):
    total=0
    for no in no:
       total+=no
    return total
no=[10,20,30,40]
print(find_sum(no))



```

```python
#function with dictionary
def dis_std(std):
    for key,value in std.items():
        print(key,":",value)
std ={
    "name":"rahul",
    "age":22,
    "course":"python"
}
dis_std(std)
```

```python
#function calling another function
def square(no):
    return no*no
def dis_square(no):
    result=square(no)
    print("square:",result)
dis_square(245)
```

```python
#function with condition
def che_marks(marks):
    if marks>=40:
        return"pass"
    else:
        return"fail"
print(che_marks(70))
print(che_marks(30))
```

```python
#function with loop 
def print_no(n):
    for i in range(1,n+1):
     print(i)
print_no(4)
```

```python
#function for factorial
def fac(n):
    result=10
    for i in range(1,n+1):
        result=result*1
    return result
print(fac(10))
```

```python
#function for largest number
def find_largest(no):
    largest=no[0]
    for no in no:
     if no>largest:
        largest=no
    return largest
no=[10,20,30,40]
print(find_largest(no)) 
```
